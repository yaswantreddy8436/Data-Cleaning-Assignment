```python
1 print(df.isnull().sum())

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)

2 df["Age"] = df["Age"].fillna(df["Age"].median())

print(df["Age"].isnull().sum())

# Median is used because Age may contain outliers.

3 df["Treatment_Cost"] = df["Treatment_Cost"].fillna(df["Treatment_Cost"].median())

print(df["Treatment_Cost"].isnull().sum())

# Median is used because Treatment_Cost is highly skewed.

4  before = df.shape[0]

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

df = df.drop_duplicates()

after = df.shape[0]

print("Before:", before)

print("After:", after)


5 invalid_age = df[(df["Age"] < 0) | (df["Age"] > 100)]

print(invalid_age)

df = df[(df["Age"] >= 0) & (df["Age"] <= 100)]

print("Invalid age records removed.")



6 Q1 = df["Treatment_Cost"].quantile(0.25)

Q3 = df["Treatment_Cost"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[(df["Treatment_Cost"] < lower) | (df["Treatment_Cost"] > upper)]

print("Number of Outliers:", len(outliers))


7 lower = df["Treatment_Cost"].quantile(0.05)

upper = df["Treatment_Cost"].quantile(0.95)

df["Treatment_Cost"] = df["Treatment_Cost"].clip(lower, upper)

print(df["Treatment_Cost"].describe())


8 import numpy as np

df["Log_Treatment_Cost"] = np.log1p(df["Treatment_Cost"])

print(df[["Treatment_Cost", "Log_Treatment_Cost"]].head())

print("Before Distribution")

print(df["Treatment_Cost"].describe())

print("After Distribution")

print(df["Log_Treatment_Cost"].describe())



9 df = df.sort_values("Admission_Date")

df["Admission_Date"] = df["Admission_Date"].fillna(method="ffill")

print(df["Admission_Date"].head())

# Forward fill is used because admission dates follow a chronological sequence.
```
